```python
#Write a Python program to calculate the sum of three given numbers, if the values are equal then return thrice of their sum.
```

```python
x = int(input("Enter x: "))
y = int(input("Enter y: "))
z = int(input("Enter z: "))

if x == y == z:
    print(3 * x)
else:
    print(x + y + z)
```

```python
#Write a Python program to print all even numbers from 1 to n using for loop.
```

```python
n = int(input("Enter n: "))

for i in range(1, n + 1):
    if i % 2 == 0:
        print(i)
```

```python
#Python Program to Find the Factorial of a Number
```

```python
num = int(input("Enter a number: "))
factorial = 1

if num < 0:
    print("Factorial does not exist for negative numbers")

elif num == 0:
    print("The factorial of 0 is 1")

else:
    for i in range(1, num + 1):
        factorial *= i

    print("Factorial:", factorial)
```

```python
#Basic Image Processing and Analysis Using Python
```

```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

# Read image
image = Image.open("a.jpg")#.convert("L")

# Display image
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.show()

# Image information
print("Image size:", image.size)
print("Image mode:", image.mode)

# Convert image to NumPy array
img_array = np.array(image)

print("Array shape:", img_array.shape)
print("Data type:", img_array.dtype)

# Access a pixel
pixel = img_array[100, 100]
print("Pixel value:", pixel)

# Convert to grayscale
gray = image.convert("L")

plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()

print("Grayscale shape:", np.array(gray).shape)
```

```python
#Compare PIL and OpenCV Grayscale Images
```

```python
img_pil = Image.open("a.jpg")
img_cv = cv2.imread("b.jpg")

# Convert to grayscale
img_gray_pil = img_pil.convert("L")
img_gray_cv = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)

# Convert PIL image to NumPy array
img_gray_pil_np = np.array(img_gray_pil)

# Display both images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_gray_pil_np, cmap="gray")
plt.title("PIL Grayscale")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_gray_cv, cmap="gray")
plt.title("OpenCV Grayscale")
plt.axis("off")

plt.show()
```
