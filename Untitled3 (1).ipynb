```python
# 1. Display a Grayscale Image and Its Histogram
# Purpose: Load an image, convert it to grayscale, and display its histogram to show pixel intensity distribution.
```

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Load the image and convert it to grayscale
image = Image.open(r"a.jpg").convert("L")
image_array = np.array(image)

# Create a figure with two subplots
plt.figure(figsize=(12, 6))

# Display the original grayscale image
plt.subplot(1, 3, 1)
plt.imshow(image_array, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')

# Display the image histogram
plt.subplot(1, 3, 3)
plt.hist(image_array.ravel(), bins=256, range=[0, 256], color='gray')
plt.title('Histogram of Pixel Intensity')
plt.xlabel('Pixel Intensity (0-255)')
plt.ylabel('Frequency')

# Display the results
plt.show()
```

```python
# 2. Apply Negative Transformation
# Purpose: Convert a grayscale image into its negative by subtracting each pixel intensity from 255.
```

```python
# Import required libraries
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

# 1. Load the image and convert it to grayscale
img_gray = Image.open(r"a.jpg").convert("L")  # "L" = 8-bit grayscale

# 2. Convert the grayscale image into an array
gray_arr = np.array(img_gray)

# 3. Apply Negative Transformation
negative_gray = 255 - (gray_arr)

# 4. Display the results
plt.figure(figsize=(12, 4))

# Display the original grayscale image
plt.subplot(1, 2, 1)
plt.imshow(img_gray, cmap="gray")
plt.title("Original Gray")
plt.axis("off")

# Display the negative image
plt.subplot(1, 2, 2)
plt.imshow(negative_gray, cmap='gray')
plt.title("Negative Gray (PIL)")
plt.axis("off")

plt.show()
```

```python
# 3. Apply Four Contrast Enhancement Techniques
# Purpose: Apply contrast stretching, low contrast, high contrast, and gamma correction to an image.
```

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Load the image and convert it to grayscale
x = Image.open(r"C:\Users\PC\Desktop\i.jpg").convert('L')
x_array = np.array(x)

# Contrast Stretching
x_min = np.min(x_array)
x_max = np.max(x_array)
y = ((x_array - x_min) / (x_max - x_min) * 255).astype(np.uint8)

# Low Contrast (by multiplying by a value less than 1)
z = (x_array * 0.4).clip(0, 255).astype(np.uint8)

# High Contrast (by multiplying by a value greater than 1)
w = (x_array * 3.4).clip(0, 255).astype(np.uint8)

# Apply Gamma Correction
gamma = 3.0  # Greater than 1 = darker, less than 1 = brighter
c = np.power(x_array / 255.0, gamma) * 255.0
c = np.clip(c, 0, 255).astype(np.uint8)

# Display the four contrast techniques
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.imshow(y, cmap='gray')
plt.title('Contrast Stretched Image')

plt.subplot(2, 2, 2)
plt.imshow(z, cmap='gray')
plt.title('Low Contrast')

plt.subplot(2, 2, 3)
plt.imshow(w, cmap='gray')
plt.title('High Contrast')

plt.subplot(2, 2, 4)
plt.imshow(c, cmap='gray')
plt.title('Contrast with Gamma')

plt.show()
```

```python
# 4. Apply Histogram Equalization
# Purpose: Improve image contrast using histogram equalization and compare the original and equalized histograms.
```

```python
# Import required libraries
from PIL import Image, ImageOps
import matplotlib.pyplot as plt
import numpy as np

# 1. Load the image and convert it to grayscale
img = Image.open("a.jpg").convert("L")

# 2. Apply histogram equalization
equalized_img = ImageOps.equalize(img)

# 3. Display original and equalized images
plt.figure(figsize=(12, 12))

# Display the original image
plt.subplot(2, 2, 1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')

# Display the original image histogram
plt.subplot(2, 2, 2)
plt.hist(np.array(img).ravel(), bins=256, range=[0, 256], color='gray')
plt.title('Histogram of Original Image')
plt.xlabel('Pixel Intensity (0-255)')
plt.ylabel('Frequency')

# Display the histogram equalized image
plt.subplot(2, 2, 3)
plt.title("Histogram Equalized Image")
plt.imshow(equalized_img, cmap='gray')

# Display the equalized image histogram
plt.subplot(2, 2, 4)
plt.hist(np.array(equalized_img).ravel(), bins=256, range=[0, 256], color='gray')
plt.title('Histogram of Equalized Image')
plt.xlabel('Pixel Intensity (0-255)')
plt.ylabel('Frequency')

plt.show()
```

```python
#5. Apply Mean (Average) Filter Manually Using Convolve
#Purpose: Add salt-and-pepper noise to an image and reduce it using a manually defined 3×3 mean filter and the convolution function.
```

```python
# Import required libraries
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import convolve
from skimage.util import random_noise
from skimage import img_as_ubyte

# Load the grayscale image
img = Image.open("cameraman.png").convert("L")
img_array = np.array(img)

# Add salt-and-pepper noise
noisy_image_array = random_noise(img_array, mode='s&p', amount=0.02)
noisy_image = img_as_ubyte(noisy_image_array)

# Define a 3x3 mean filter
kernel = np.ones((3, 3)) / 9.0

# Apply convolution (filtering)
mean_filtered = convolve(noisy_image, kernel)

# Plot Original, Noisy, and Mean Filtered images
plt.figure(figsize=(12, 12))

# Display the original image
plt.subplot(131)
plt.imshow(img_array, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# Display the noisy image
plt.subplot(132)
plt.imshow(noisy_image, cmap='gray')
plt.title('Image with Salt-and-Pepper Noise')
plt.axis('off')

# Display the mean filtered image
plt.subplot(133)
plt.imshow(mean_filtered, cmap='gray')
plt.title('Mean Filtered Image')
plt.axis('off')

plt.show()
```

```python
#6. Apply Mean (Uniform) Filter Using Built-in Function
#Purpose: Remove salt-and-pepper noise using the built-in uniform_filter() function.
```

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from scipy.ndimage import uniform_filter
from skimage.util import random_noise
from skimage import img_as_ubyte

# Load the image and convert it to grayscale
image = Image.open('a.jpg').convert('L')
image_array = np.array(image)

# Add salt-and-pepper noise
noisy_image_array = random_noise(image_array, mode='s&p', amount=0.02)
noisy_image = img_as_ubyte(noisy_image_array)

# Apply Mean (Average) Filter
mean_filtered = uniform_filter(noisy_image, size=3)

# Plot Original, Noisy, and Mean Filtered images
plt.figure(figsize=(12, 12))

# Display the original image
plt.subplot(131)
plt.imshow(image_array, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# Display the noisy image
plt.subplot(132)
plt.imshow(noisy_image, cmap='gray')
plt.title('Image with Salt-and-Pepper Noise')
plt.axis('off')

# Display the mean filtered image
plt.subplot(133)
plt.imshow(mean_filtered, cmap='gray')
plt.title('Mean Filtered Image')
plt.axis('off')

plt.show()
```

```python
#7. Apply Median Filter
#Purpose: Add salt-and-pepper noise to an image and remove it using the median filter.
```

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from scipy.ndimage import (
    median_filter, uniform_filter, maximum_filter, minimum_filter
)
from skimage.util import random_noise
from skimage import img_as_ubyte

# Load the image and convert it to grayscale
image = Image.open('a.jpg').convert('L')
image_array = np.array(image)

# Add salt-and-pepper noise
noisy_image_array = random_noise(
    image_array, mode='s&p', amount=0.5
)
noisy_image = img_as_ubyte(noisy_image_array)

# Apply Median Filter
median_filtered = median_filter(noisy_image, size=5)

# Plot the original, noisy, and filtered images
plt.figure(figsize=(15, 12))

# Display the original image
plt.subplot(1, 3, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# Display the noisy image
plt.subplot(1, 3, 2)
plt.imshow(noisy_image, cmap='gray')
plt.title('Image with Salt-and-Pepper Noise')
plt.axis('off')

# Display the median filtered image
plt.subplot(1, 3, 3)
plt.imshow(median_filtered, cmap='gray')
plt.title('Median Filtered Image')
plt.axis('off')

plt.show()
```

```python
#8. Apply Maximum Filter
#Purpose: Apply the maximum filter to expand bright features and demonstrate its effect on noisy and original images.
```

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from scipy.ndimage import (
    median_filter, uniform_filter, maximum_filter, minimum_filter
)
from skimage.util import random_noise
from skimage import img_as_ubyte

# Load the image and convert it to grayscale
image = Image.open('a.jpg').convert('L')
image_array = np.array(image)

# Add salt-and-pepper noise
noisy_image_array = random_noise(
    image_array, mode='s&p', amount=0.02
)
noisy_image = img_as_ubyte(noisy_image_array)

# Apply Maximum Filter
max_filtered_with_noise = maximum_filter(noisy_image, size=5)
max_without_noise = maximum_filter(image, size=5)

# Plot the original, noisy, and filtered images
plt.figure(figsize=(15, 12))

# Display the original image
plt.subplot(1, 4, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# Display the noisy image
plt.subplot(1, 4, 2)
plt.imshow(noisy_image, cmap='gray')
plt.title('Image with Salt-and-Pepper Noise')
plt.axis('off')

# Display the maximum filtered image with noise
plt.subplot(1, 4, 3)
plt.imshow(max_filtered_with_noise, cmap='gray')
plt.title('Maximum Filtered Image with Noise')
plt.axis('off')

# Display the maximum filtered image without noise
plt.subplot(1, 4, 4)
plt.imshow(max_without_noise, cmap='gray')
plt.title('Maximum Filtered Image without Noise\n(Edge Detection)')
plt.axis('off')

# Adjust the padding of the subplots
plt.tight_layout()

# Display the plots
plt.show()
```
