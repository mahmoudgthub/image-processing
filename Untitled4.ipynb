```python
# Import required libraries
from PIL import Image
import numpy as np
from scipy.ndimage import convolve
import matplotlib.pyplot as plt

# Load the image and convert it to grayscale
image = Image.open(r'a.jpg').convert('L')

# Convert image to float to support negative values
float_image = np.float32(image)

# Define Prewitt kernels
# Horizontal edge detection (Y-gradient)
ph = np.array([[-1, -1, -1],
               [ 0,  0,  0],
               [ 1,  1,  1]])

# Vertical edge detection (X-gradient)
pv = np.array([[-1,  0,  1],
               [-1,  0,  1],
               [-1,  0,  1]])

# Apply convolution to detect edges
h = convolve(float_image, ph)
v = convolve(float_image, pv)

# Calculate the final gradient magnitude
magnitude = np.sqrt(h**2 + v**2).clip(0, 255).astype(np.uint8)

# Display the results
plt.figure(figsize=(12, 10))

# Display horizontal edges
plt.subplot(2, 2, 1)
plt.imshow(h, cmap='gray')
plt.title("The Horizontal Edges (Y-Gradient) Image")

# Display vertical edges
plt.subplot(2, 2, 2)
plt.imshow(v, cmap='gray')
plt.title("The Vertical Edges (X-Gradient) Image")

# Display the final gradient image
plt.subplot(2, 2, (3, 4))
plt.imshow(magnitude, cmap='gray')
plt.title("The Final Gradient Image")

# Hide axes and adjust the layout
for ax in plt.gcf().get_axes():
    ax.axis('off')

plt.tight_layout()
```

```python
#2. Sobel Operator
#Page 3–4 | Horizontal, Vertical & Diagonal Edge Detection
```

```python
# Import required libraries
from PIL import Image
import numpy as np
from scipy.ndimage import convolve
import matplotlib.pyplot as plt

# Load the image and convert it to grayscale
image = Image.open(r'a.jpg').convert('L')

# Convert image to float to support negative values
float_image = np.float32(image)

# Define Sobel kernels

# Horizontal edge detection (Y-gradient)
sh = np.array([[-1, -2, -1],
               [ 0,  0,  0],
               [ 1,  2,  1]])

# Vertical edge detection (X-gradient)
sv = np.array([[-1,  0,  1],
               [-2,  0,  2],
               [-1,  0,  1]])

# First diagonal edge detection
sd1 = np.array([[ 0,  1,  2],
                [-1,  0,  1],
                [-2, -1,  0]])

# Second diagonal edge detection
sd2 = np.array([[-2, -1,  0],
                [-1,  0,  1],
                [ 0,  1,  2]])

# Apply convolution to detect edges
h = convolve(float_image, sh)
v = convolve(float_image, sv)
d1 = convolve(float_image, sd1)
d2 = convolve(float_image, sd2)

# Calculate horizontal and vertical gradient magnitude
magnitude_hv = np.sqrt(h**2 + v**2).clip(0, 255).astype(np.uint8)

# Calculate diagonal gradient magnitude
magnitude_d1d2 = np.sqrt(d1**2 + d2**2).clip(0, 255).astype(np.uint8)

# Display the results
plt.figure(figsize=(12, 13))

# Display horizontal edges
plt.subplot(2, 3, 1)
plt.imshow(h, cmap='gray')
plt.title("The Horizontal Edges (Y-Gradient) Image")

# Display vertical edges
plt.subplot(2, 3, 2)
plt.imshow(v, cmap='gray')
plt.title("The Vertical Edges (X-Gradient) Image")

# Display final horizontal and vertical gradient
plt.subplot(2, 3, 3)
plt.imshow(magnitude_hv, cmap='gray')
plt.title("The Final Gradient Image")

# Display first diagonal edges
plt.subplot(2, 3, 4)
plt.imshow(d1, cmap='gray')
plt.title(r"The 1st Diagonal Edges \ Image")

# Display second diagonal edges
plt.subplot(2, 3, 5)
plt.imshow(d2, cmap='gray')
plt.title("The 2nd Diagonal Edges / Image")

# Display final diagonal gradient
plt.subplot(2, 3, 6)
plt.imshow(magnitude_d1d2, cmap='gray')
plt.title("The Final Gradient Image")

# Hide axes and adjust the layout
for ax in plt.gcf().get_axes():
    ax.axis('off')

plt.tight_layout()
```
